```python
import os

class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

    def display_info(self):
        print(f" 👤 Name   : {self.name}")
        print(f" 🎂 Age    : {self.age}")
        print(f" 📚 Course : {self.course}")


students = []
unsaved_changes = False

def load_students():
    global students, unsaved_changes
    students = [] # Reset/initialize list
    if not os.path.exists("students.txt"):
        # If file doesn't exist yet, we start with our default students
        student1 = Student("Roline", 20, "Data Science")
        student2 = Student("John", 21, "Computer Science")
        students.append(student1)
        students.append(student2)
        return

    try:
        with open("students.txt", "r") as file:
            for line in file:
                line = line.strip()
                if not line:
                    continue
                parts = line.split(",")
                if len(parts) == 3:
                    name, age_str, course = parts
                    try:
                        age = int(age_str)
                        # Avoid duplicates while loading from file
                        if not any(s.name.strip().lower() == name.strip().lower() for s in students):
                            students.append(Student(name.strip(), age, course.strip()))
                    except ValueError:
                        continue
        print("\n[✓] Loaded student records from local database file successfully!")
        unsaved_changes = False
    except Exception as e:
        print(f"\n[✗] An error occurred while loading: {e}")

def add_student():
    global unsaved_changes
    print("\n--- ➕ Add New Student ---")
    name = input("Enter student's full name: ").strip()
    if not name:
        print("[!] Error: Name cannot be empty!")
        return

    # Strict check for duplicates (case-insensitive and stripping any stray whitespace)
    if any(student.name.strip().lower() == name.lower() for student in students):
        print(f"[!] Error: A student named '{name}' already exists in the registry!")
        return

    try:
        age = int(input("Enter age: "))
        if age < 0:
            print("[!] Error: Age cannot be a negative value!")
            return
    except ValueError:
        print("[!] Error: Invalid age. Please enter a valid integer.")
        return

    course = input("Enter enrolled course: ").strip()
    if not course:
        print("[!] Error: Course field cannot be empty!")
        return

    student = Student(name, age, course)
    students.append(student)
    unsaved_changes = True
    save_students()
    print(f"[✓] Student '{name}' successfully registered!")


def search_student():
    print("\n--- 🔍 Search Records ---")
    name = input("Enter student name to look up: ").strip()

    for student in students:
        if student.name.strip().lower() == name.lower():
            print("\nMatch Found:")
            print("=" * 30)
            student.display_info()
            print("=" * 30)
            return

    print("[!] No matching student found with that name.")


def delete_student():
    global unsaved_changes
    print("\n" + "=" * 40)
    print("          DELETE STUDENT RECORD")
    print("=" * 40)

    if not students:
        print("[!] No student records are available to delete.")
        return

    name = input("Enter student name to remove: ").strip()

    if not name:
        print("[!] Error: Student name cannot be empty.")
        return

    for student in students:
        if student.name.strip().lower() == name.lower():

            print("\nStudent record found:")
            print("-" * 30)
            student.display_info()
            print("-" * 30)

            confirm = input(
                "Are you sure you want to delete this student? (yes/no): "
            ).strip().lower()

            if confirm == "yes":
                students.remove(student)
                unsaved_changes = True
                save_students()
                print(
                    f"\n[✓] Student '{student.name}' "
                    "has been removed successfully."
                )


            elif confirm == "no":
                print("\n[!] Deletion cancelled.")

            else:
                print("\n[!] Invalid response. Deletion cancelled.")

            return

    print(f"\n[!] No student named '{name}' was found.")

def update_student():
    global unsaved_changes
    print("\n--- 📝 Update Student Details ---")
    name = input("Enter student name to modify: ").strip()

    for student in students:
        if student.name.strip().lower() == name.lower():
            print(f"\nUpdating record for '{student.name}':")
            try:
                new_age = int(input("Enter new age: "))
                if new_age < 0:
                    print("[!] Error: Age cannot be negative!")
                    return
                student.age = new_age
            except ValueError:
                print("[!] Error: Invalid numerical entry. Update aborted.")
                return

            new_course = input("Enter new course (or press Enter to keep current): ").strip()
            if new_course:
                student.course = new_course
            unsaved_changes = True
            save_students()
            print("[✓] Record successfully updated!")
            return

    print("[!] Student record not found.")

def save_students():
    global unsaved_changes
    print("\n--- 💾 Saving Database ---")
    try:
        with open("students.txt", "w") as file:
            for student in students:
                file.write(
                    f"{student.name.strip()},{student.age},{student.course.strip()}\n"
                )
        unsaved_changes = False
        print("[✓] All student records have been saved securely.")
        return True
    except Exception as e:
        unsaved_changes = True
        print(f"[✗] An error occurred while saving: {e}")
        return False

# Initialize student database at launch
load_students()

print("\n")
print("=" * 45)
print("   WELCOME TO THE STUDENT MANAGEMENT SYSTEM")
print("=" * 45)
print("Manage student records easily and efficiently.")
print()

while True:
    print("\n" + "=" * 40)
    print("    🎓  STUDENT REGISTRY SYSTEM  🎓")
    print("=" * 40)
    print("  [1]  Add New Student")
    print("  [2]  View All Enrolled Students")
    print("  [3]  Search Student Record")
    print("  [4]  Remove Student Record")
    print("  [5]  Update Existing Details")
    print("  [6]  Save Changes to Disk")
    print("  [7]  Reload Local Database File")
    print("  [8]  Exit Program")
    print("=" * 40)

    choice = input("👉 Please select an action (1-8): ").strip()

    if choice == "1":
        add_student()

    elif choice == "2":
        print("\n--- 📋 Student Roster ---")
        if not students:
            print("The student registry is currently empty.")
        else:
            for i, student in enumerate(students, 1):
                print(f"Record #{i}:")
                student.display_info()
                print("-" * 30)

    elif choice == "3":
        search_student()

    elif choice == "4":
        delete_student()

    elif choice == "5":
        update_student()

    elif choice == "6":
        save_students()

    elif choice == "7":
        if unsaved_changes:
            confirm = input(
                "You have unsaved changes. Reload anyway? (yes/no): "
            ).strip().lower()
            if confirm != "yes":
                print("[i] Reload cancelled.")
                continue
        load_students()

    elif choice == "8":
        if unsaved_changes:
            confirm = input(
                "You have unsaved changes. Exit anyway? (yes/no): "
            ).strip().lower()
            if confirm != "yes":
                print("[i] Exit cancelled.")
                continue
        print("\nThank you for using Student Registry. Goodbye!")
        break

    else:
        print("\n[!] Invalid Selection. Please enter a valid number between 1 and 8.")
```
